## 0.5 — Key Mathematical Concepts Preview

This table previews every major formula you'll encounter. Use it as a quick reference throughout the course.

| Concept | Formula | Notebook | Purpose |
|---|---|---|---|
| **EMA** | $\text{EMA}_k(t) = \alpha \cdot P_t + (1-\alpha) \cdot \text{EMA}_k(t-1), \quad \alpha = \frac{2}{k+1}$ | 02 | Smoothed price trend |
| **RSI** | $\text{RSI} = 100 - \frac{100}{1 + \text{AvgGain}/\text{AvgLoss}}$ | 02, 04, 05 | Overbought/oversold detection |
| **Bollinger Bands** | $\text{BB}_{\text{lower}} = \text{SMA}_{20} - 2\sigma_{20}$ | 02, 05 | Volatility envelope |
| **Momentum Score** | $M_i(t) = \frac{1}{|K|} \sum_{L \in K} \left(\frac{P_{i,t}}{P_{i,t-L}} - 1\right)$ | 04 | Multi-lookback return |
| **Mean-Rev Signal** | $\max\left(\text{Sig}_{\text{RSI}},\; \text{Sig}_{\text{BB}}\right)$ | 05 | Composite oversold signal |
| **Hedge Ratio** | $A = \beta B + \epsilon \;\;(\text{OLS regression})$ | 06 | Pairs trading spread |
| **Half-Life** | $t_{1/2} = -\ln(2) / \beta_{\text{AR}(1)}$ | 06 | Mean-reversion speed |
| **Regime** | EMA crossover + volatility threshold → Bull / Ranging / Bear | 07 | Market state classification |
| **Inv-Vol Weight** | $w_i = \frac{S_i / \hat{\sigma}_i}{\sum_j S_j / \hat{\sigma}_j}$ | 08 | Position sizing |
| **Half-Kelly** | $f^* = \frac{p \cdot b - q}{b}, \quad \text{use } f^*/2$ | 08 | Maximum position cap |
| **Sharpe** | $\frac{\bar{r} - r_f}{\sigma_r} \cdot \sqrt{A}$ | 03, 09 | Risk-adjusted return |
| **Sortino** | $\frac{\bar{r} - T}{\text{DD}} \cdot \sqrt{A}$ | 03, 09 | Downside-focused metric |
| **Calmar** | $\frac{R_{\text{ann}}}{|\text{MaxDD}|}$ | 03, 09 | Worst-loss resilience |

## 0.9 — Glossary of Key Terms

Quick reference for the ~30 most important terms. Each links to the notebook where it's first introduced.

| Term | Definition | First seen |
|---|---|---|
| **OHLCV** | Open, High, Low, Close, Volume — the 5 values in a price candle | NB 01 |

| **Sortino Ratio** | Excess return per unit of downside risk | NB 03 |
| **Calmar Ratio** | Annualized return divided by maximum drawdown | NB 03 |
| **Drawdown** | Peak-to-trough decline in portfolio value | NB 03 |
| **Momentum** | Tendency of recent winners to keep winning | NB 04 |
| **Composite Score** | Multi-lookback averaged return used for ranking | NB 04 |
| **Mean Reversion** | Tendency of extreme moves to reverse toward the mean | NB 05 |
| **Oversold** | Asset whose RSI is below a threshold (e.g., 30) | NB 05 |

| **Hedge Ratio** | Regression slope used to construct a pairs trading spread | NB 06 |

| **Cash Floor** | Minimum fraction held as cash (varies by regime) | NB 08 |
| **Kelly Criterion** | Optimal bet size based on win probability and payoff ratio | NB 08 |
| **Circuit Breaker** | Drawdown threshold that triggers position reduction or trading halt | NB 08 |
| **Rebalance** | Adjusting positions to match target weights | NB 08 |
