## 0.5 — 核心数学概念预览

下表预览了你将遇到的所有主要公式。学习过程中可随时参考。

| 概念 | 公式 | 课程 | 用途 |
|---|---|---|---|
| **EMA**（指数移动平均） | $\text{EMA}_k(t) = \alpha \cdot P_t + (1-\alpha) \cdot \text{EMA}_k(t-1), \quad \alpha = \frac{2}{k+1}$ | 02 | 平滑价格趋势 |
| **RSI**（相对强弱指数） | $\text{RSI} = 100 - \frac{100}{1 + \text{平均涨幅}/\text{平均跌幅}}$ | 02, 04, 05 | 超买/超卖检测 |
| **布林带** | $\text{BB}_{\text{lower}} = \text{SMA}_{20} - 2\sigma_{20}$ | 02, 05 | 波动率通道 |
| **动量分数** | $M_i(t) = \frac{1}{|K|} \sum_{L \in K} \left(\frac{P_{i,t}}{P_{i,t-L}} - 1\right)$ | 04 | 多周期平均回报 |
| **均值回归信号** | $\max\left(\text{Sig}_{\text{RSI}},\; \text{Sig}_{\text{BB}}\right)$ | 05 | 综合超卖信号 |
| **对冲比率** | $A = \beta B + \epsilon \;\;(\text{OLS 回归})$ | 06 | 配对交易价差 |
| **半衰期** | $t_{1/2} = -\ln(2) / \beta_{\text{AR}(1)}$ | 06 | 均值回归速度 |
| **制度分类** | EMA 交叉 + 波动率阈值 → 牛市 / 震荡 / 熊市 | 07 | 市场状态判定 |
| **反波动率加权** | $w_i = \frac{S_i / \hat{\sigma}_i}{\sum_j S_j / \hat{\sigma}_j}$ | 08 | 仓位分配 |
| **半 Kelly 约束** | $f^* = \frac{p \cdot b - q}{b}, \quad \text{使用 } f^*/2$ | 08 | 最大仓位上限 |
| **Sharpe 比率** | $\frac{\bar{r} - r_f}{\sigma_r} \cdot \sqrt{A}$ | 03, 09 | 风险调整收益 |
| **Sortino 比率** | $\frac{\bar{r} - T}{\text{DD}} \cdot \sqrt{A}$ | 03, 09 | 仅惩罚下行风险 |
| **Calmar 比率** | $\frac{R_{\text{ann}}}{|\text{MaxDD}|}$ | 03, 09 | 最大回撤韧性 |
